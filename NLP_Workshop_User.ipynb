**Example Regression vs SVM**

```python
#JUST EXAMPLE TO SEE VISUAL DIFFERENCE / IGNORE FOR MODEL
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
X, y = make_classification(n_samples=100, n_features=2, n_redundant=0,
                           n_clusters_per_class=1, class_sep=0.8, random_state=42)

lr  = LogisticRegression().fit(X, y)
svm = SVC(kernel='linear', C=1.0).fit(X, y)  # SVC instead of LinearSVC so we can draw margins

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Logistic Regression vs Linear SVM', fontsize=14, fontweight='bold')

colors = ['#e05c5c', '#4a9e6b']

# ── LEFT: Logistic Regression ────────────────────────────────────
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300),
                     np.linspace(y_min, y_max, 300))

Z_lr = lr.predict_proba(np.c_[xx.ravel(), yy.ravel()])[:, 1].reshape(xx.shape)
axes[0].contourf(xx, yy, Z_lr, levels=20, cmap='RdYlGn', alpha=0.4)
axes[0].contour(xx, yy, Z_lr, levels=[0.5], colors='navy', linewidths=2)
axes[0].scatter(X[:, 0], X[:, 1], c=[colors[i] for i in y], edgecolors='k', s=50, zorder=3)
axes[0].set_title('Logistic Regression\nsoft gradient — predicts probabilities', fontsize=11, fontweight='bold')
axes[0].set_xlabel('Feature 1')
axes[0].set_ylabel('Feature 2')

# ── RIGHT: SVM with margin lines ─────────────────────────────────
Z_svm = svm.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
axes[1].contourf(xx, yy, Z_svm, alpha=0.15, cmap='RdYlGn')

# Decision boundary + margin lines from support vectors
w  = svm.coef_[0]
b  = svm.intercept_[0]
x_line = np.linspace(x_min, x_max, 300)

axes[1].plot(x_line, -(w[0] * x_line + b) / w[1],
             color='navy', linewidth=2.5, label='decision boundary')
axes[1].plot(x_line, -(w[0] * x_line + b - 1) / w[1],
             color='gray', linewidth=1.5, linestyle='--', label='margin')
axes[1].plot(x_line, -(w[0] * x_line + b + 1) / w[1],
             color='gray', linewidth=1.5, linestyle='--')

# Highlight support vectors
axes[1].scatter(svm.support_vectors_[:, 0], svm.support_vectors_[:, 1],
                s=180, facecolors='none', edgecolors='gold',
                linewidths=2.5, zorder=4, label='support vectors')
axes[1].scatter(X[:, 0], X[:, 1], c=[colors[i] for i in y], edgecolors='k', s=50, zorder=3)

axes[1].set_title('Linear SVM\nhard boundary — maximizes the margin', fontsize=11, fontweight='bold')
axes[1].set_xlabel('Feature 1')
axes[1].set_ylabel('Feature 2')
axes[1].legend(loc='upper left', fontsize=9)

plt.tight_layout()
plt.show()
```
